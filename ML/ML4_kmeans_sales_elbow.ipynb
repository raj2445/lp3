```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

print("All libraries imported successfully!")
```

```python
DATA_FILE = "sales_data_sample.csv"

df = pd.read_csv(DATA_FILE, encoding="latin1")

print("Dataset loaded successfully!")

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# Select numeric columns
X = df.select_dtypes(include="number").copy()

# Remove columns that are likely identifiers
X = X.drop(
    columns=[
        c for c in X.columns
        if any(s in c.lower()
               for s in ["id", "number", "postal", "phone"])
    ],
    errors="ignore"
)

if X.shape[1] == 0:
    raise ValueError("No suitable numeric columns found!")

print("Selected Numerical Features:")
print(X.columns.tolist())

print("\nFeature Matrix Shape:")
print(X.shape)
```

```python
imputer = SimpleImputer(strategy="median")

X = pd.DataFrame(
    imputer.fit_transform(X),
    columns=X.columns
)

print("Missing values handled successfully!")

print("\nRemaining Missing Values:")
print(X.isnull().sum().sum())
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Feature scaling completed!")

print("Scaled Data Shape:", X_scaled.shape)
```

```python
inertias = []

ks = range(1, 11)

for k in ks:

    km = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    km.fit(X_scaled)

    inertias.append(km.inertia_)

print("Inertia values:")
print(inertias)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    list(ks),
    inertias,
    marker="o"
)

plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")

plt.title("Elbow Method")

plt.xticks(list(ks))
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
k = int(input("Choose k after inspecting the elbow plot: "))

if k < 2 or k > 10:
    raise ValueError("Please choose k between 2 and 10.")

print("Selected Number of Clusters:", k)
```

```python
model = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)

clusters = model.fit_predict(X_scaled)

df["Cluster"] = clusters

print("K-Means clustering completed successfully!")
```

```python
print("Customer Cluster Distribution:")

print(
    df["Cluster"]
    .value_counts()
    .sort_index()
)
```

```python
print("Final Dataset with Cluster Labels:")

display(df.head(10))

print("\nCluster Summary:")

display(
    df.groupby("Cluster")[X.columns].mean().round(2)
)
```
